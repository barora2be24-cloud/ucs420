```python
import pandas as pd

roll_number = "1024170431"

fixed_entries = [{'question': 'what is the annual fee', 'answer': 'The annual fee is Rs 500.', 'keywords': 'fee cost price charge', 'category': 'billing'}, {'question': 'how to reset password', 'answer': 'Go to Settings > Reset Password.', 'keywords': 'password reset login', 'category': 'account'}, {'question': 'what are your working hours', 'answer': 'We are open 9 AM to 5 PM.', 'keywords': 'hours timing open time', 'category': 'general'}, {'question': 'how can i pay the fee', 'answer': 'You can pay via UPI, card, or net banking.', 'keywords': 'pay payment upi fee', 'category': 'billing'}]

last_two_digits = [int(roll_number[-2]), int(roll_number[-1])]
category_list = ["billing", "account", "general"]

personalized_entries = [{'question': 'how can i pay my semester fee online', 'answer': 'You can pay the semester fee online using UPI, card, or net banking.', 'keywords': 'semester fee online pay', 'category': 'billing'}, {'question': 'how can i update my registered mobile number', 'answer': 'Go to Account Settings and choose the option to update your registered mobile number.', 'keywords': 'mobile number update', 'category': 'account'}]

faq_entries = fixed_entries + personalized_entries
df = pd.DataFrame(faq_entries)

print("Last two digits:", last_two_digits)
print("\nFinal 6-row DataFrame:")
display(df)
```

 Q2: 

```python
def score_faq(query, df):
    query_words = set(query.lower().split())
    results = []

    for index, row in df.iterrows():
        searchable_words = set(row["question"].lower().split()) | set(row["keywords"].lower().split())
        score = len(query_words & searchable_words)

        if score > 0:
            results.append({
                "index": index,
                "question": row["question"],
                "answer": row["answer"],
                "category": row["category"],
                "score": score
            })

    if not results:
        return pd.DataFrame()

    return pd.DataFrame(results).sort_values("score", ascending=False).reset_index(drop=True)

query = "how can i pay the fee"
print("Query:", query)
display(score_faq(query, df))
```
