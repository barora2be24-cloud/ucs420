```python
import io

emp_data = """Employee_ID,Name,Department,Age,Salary,Years_of_Experience,Joining_Date,Gender,Bonus,Rating
101,Alice,HR,29,50000,4,2020-03-15,Female,5000,4.5
102,Bob,IT,34,70000,8,2017-07-19,Male,7000,4.0
103,Charlie,IT,41,65000,10,2013-06-01,Male,6000,3.8
104,Diana,Marketing,28,55000,3,2021-02-10,Female,4500,4.7
105,Edward,Sales,38,60000,12,2010-11-25,Male,5000,3.5"""

emp_df = pd.read_csv(io.StringIO(emp_data))
emp_df.to_csv("employees.csv", index=False)
print("--- Question 6: employees.csv created successfully ---")
print(emp_df)

```

6(E)

```python
print("--- Q.6(e) Summary Statistics ---")
print(f"i. Average Salary: ${emp_df['Salary'].mean():.2f}")
print(f"ii. Total Bonus Paid: ${emp_df['Bonus'].sum():.2f}")
print(f"iii. Youngest Employee Age: {emp_df['Age'].min()} years")
print(f"iv. Highest Performance Rating: {emp_df['Rating'].max()}")
```

6(F)

```python
print("--- Q.6(f) Sorted by Salary (Descending) ---")
print(emp_df.sort_values(by="Salary", ascending=False))
```

6(G)

```python
def categorize_rating(r):
  if r >= 4.5:
    return "Excellent"
  elif r >= 4.0:
    return "Good"
  else:
    return "Average"


emp_df["Performance_Category"] = emp_df["Rating"].apply(categorize_rating)
print("--- Q.6(g) Ratings Categorized ---")
print(emp_df[["Name", "Rating", "Performance_Category"]])
```

6(H)

```python
print("--- Q.6(h) Missing Values Count ---")
print(emp_df.isnull().sum())
```

6(I)

```python
emp_df.rename(columns={"Employee_ID": "ID"}, inplace=True)
print("--- Q.6(i) Column Renamed to 'ID' ---")
print(emp_df.columns.tolist())
```

6(J)

```python
print("--- Q.6(j) IT Employees with > 5 Years Experience ---")
filtered_emp = emp_df[
    (emp_df["Years_of_Experience"] > 5) & (emp_df["Department"] == "IT")
]
print(filtered_emp)
```

6(K)

```python
emp_df["Tax"] = emp_df["Salary"] * 0.10
print("--- Q.6(k) Tax Column Added ---")
print(emp_df[["Name", "Salary", "Tax"]])
```
