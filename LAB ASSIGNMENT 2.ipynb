```python

digits = [int(d) for d in ROLL_NO]
L = [d * 10 for d in digits[:10]]

# i. Print L
print("i. Original List L:", L)

# ii. Add two numbers using append() and insert()
num_to_append = 99
L.append(num_to_append)
print("ii. L after append():", L)

num_to_insert = 45
insert_pos = 2
L.insert(insert_pos, num_to_insert)
print("    L after insert():", L)

# iii. Remove two elements using remove() and pop()
L.remove(num_to_append)
print("iii. L after remove():", L)

popped_val = L.pop(insert_pos)
print(f"     L after pop({insert_pos}) [removed {popped_val}]:", L)

# iv. Sort L in ascending and descending order
L.sort()
print("iv. Ascending order:", L)

L.sort(reverse=True)
print("    Descending order:", L)

# v. Slicing first three and last three elements
print("v. First three elements:", L[:3])
print("   Last three elements:", L[-3:])

# vi. List comprehension for elements > average
avg_L = sum(L) / len(L)
greater_than_avg = [x for x in L if x > avg_L]
print(f"vi. Average: {avg_L:.2f}")
print("    Elements greater than average:", greater_than_avg)
```

```python
import random

ROLL_NO = "1024170431"

# 3. Set the random seed to your roll number (as integer)
random.seed(int(ROLL_NO))

# i. Generate a list of 100 random numbers between 100 and 900 (inclusive)
rand_list = [random.randint(100, 900) for _ in range(100)]
print("i. Generated 100 random numbers successfully.")

# ii. Count and print all odd numbers in the list
odd_numbers = [x for x in rand_list if x % 2 != 0]
print(f"ii. Total odd numbers: {len(odd_numbers)}")

# iii. Count and print all even numbers in the list
even_numbers = [x for x in rand_list if x % 2 == 0]
print(f"iii. Total even numbers: {len(even_numbers)}")


# iv. Count and print all prime numbers, and build list using list comprehension
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True


prime_numbers = [x for x in rand_list if is_prime(x)]
print(f"iv. Total prime numbers: {len(prime_numbers)}")
print("    List of prime numbers found:", prime_numbers)

# v. Print the number that occurs most frequently and its count
most_frequent = max(set(rand_list), key=rand_list.count)
freq_count = rand_list.count(most_frequent)
print(
    f"v. Most frequent number: {most_frequent} (occurs {freq_count} times)"
)
```

```python

digits = [int(d) for d in ROLL_NO[:8]]
A = {d * 7 for d in digits}
B = {d * 9 for d in digits}

print("Set A:", A)
print("Set B:", B)

# vi. Union
print("vi. Union (A | B):", A.union(B))

# vii. Intersection
print("vii. Intersection (A & B):", A.intersection(B))

# viii. Difference
diff_A_B = A.difference(B)
diff_B_A = B.difference(A)
print("viii. A - B:", diff_A_B)
print("      B - A:", diff_B_A)

# ix. Symmetric Difference
print("ix. Symmetric Difference:", A.symmetric_difference(B))

# x. Subset and Superset checks
print("x. Is A a subset of B?", A.issubset(B))
print("   Is B a superset of A?", B.issuperset(A))

# xi. Discard user input value X
val_X = int(input("xi. Enter a value X to remove from Set A: "))
A.discard(val_X)
print("    Set A after discard:", A)
```

```python


NAME = "Briti Arora"
ROLL_NO = "1024170431"
BRANCH = "COPC"
AGE = 20
CITY = "Patiala"
CGPA = 7.5
my_dict = {
    "name": NAME,
    "roll_no": ROLL_NO,
    "branch": BRANCH,
    "age": AGE,
    "city": CITY,
}

# i. Rename "city" to "location" using pop()
my_dict["location"] = my_dict.pop("city")
print("i. After renaming 'city' to 'location':", my_dict)

# ii. Add new key "cgpa"
my_dict["cgpa"] = CGPA
print("ii. After adding 'cgpa':", my_dict)

# iii. Update "age" by increasing it by 1
my_dict["age"] += 1
print("iii. After incrementing 'age':", my_dict)

# iv. Delete "branch" using pop() and del on separate fresh copies of my_dict
copy_for_pop = my_dict.copy()
copy_for_del = my_dict.copy()

popped_value = copy_for_pop.pop("branch")
del copy_for_del["branch"]
print(f"iv. Deleted 'branch' via pop() [returned '{popped_value}'] and del successfully.")

# v. Iterate using .items() and print "key → value"
print("v. Dictionary Key-Value pairs:")
for key, value in my_dict.items():
    print(f"   {key} → {value}")

# vi. Safe check for missing key "email" using 'in'
if "email" in my_dict:
    print("Email:", my_dict["email"])
else:
    print("vi. Safe fallback: Key 'email' does not exist in the dictionary.")

# vii. Create friend_dict with the original 5 keys and merge using **dict1, **dict2
friend_dict = {
    "name": "Riya",
    "roll_no": "1024170499",
    "branch": "COPC",
    "age": 21,
    "city": "Chandigarh",
}

merged_dict = {**my_dict, **friend_dict}
print("vii. Merged Dictionary:", merged_dict)

# viii. Dictionary comprehension for string values only
string_only_dict = {k: v for k, v in my_dict.items() if isinstance(v, str)}
print("viii. Only string key-value pairs:", string_only_dict)
```
