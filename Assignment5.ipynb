## Q2

```python
arr = np.array([1, 2, 3, 6, 4, 5])
print("Reversed array:", arr[::-1])
```

```python
x = np.array([1,2,3,4,5,1,2,1,1,1])

values, counts = np.unique(x, return_counts=True)
most_frequent = values[np.argmax(counts)]
indices = np.where(x == most_frequent)[0]

print("Most frequent value:", most_frequent)
print("Indices:", indices)
```

```python
y = np.array([1,1,1,2,3,4,2,4,3,3])

values, counts = np.unique(y, return_counts=True)
most_frequent = values[np.argmax(counts)]
indices = np.where(y == most_frequent)[0]

print("Most frequent value:", most_frequent)
print("Indices:", indices)
```

## Q4

```python
Briti_Arora = np.linspace(10, 100, 25)

print("Array:", Briti_Arora)
print("Dimensions:", Briti_Arora.ndim)
print("Shape:", Briti_Arora.shape)
print("Total elements:", Briti_Arora.size)
print("Data type:", Briti_Arora.dtype)
print("Total bytes:", Briti_Arora.nbytes)

transpose_array = Briti_Arora.reshape(25,1)
print("Transpose using reshape():")
print(transpose_array)

print("Using T:")
print(Briti_Arora.T)
```

## Q5

```python
ucs420_Briti_Arora = np.array([
    [10,20,30,40],
    [50,60,70,80],
    [90,15,20,35]
])

print("Mean:", np.mean(ucs420_Briti_Arora))
print("Median:", np.median(ucs420_Briti_Arora))
print("Max:", np.max(ucs420_Briti_Arora))
print("Min:", np.min(ucs420_Briti_Arora))
print("Unique elements:", np.unique(ucs420_Briti_Arora))

reshaped_ucs420_Briti_Arora = ucs420_Briti_Arora.reshape(4,3)
print("Reshaped array:")
print(reshaped_ucs420_Briti_Arora)

resized_ucs420_Briti_Arora = np.resize(ucs420_Briti_Arora, (2,3))
print("Resized array:")
print(resized_ucs420_Briti_Arora)
```
